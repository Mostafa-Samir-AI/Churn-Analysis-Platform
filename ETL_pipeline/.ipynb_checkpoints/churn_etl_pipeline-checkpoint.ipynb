# Telecom Churn — CSV to SQL Injection Pipeline (Microsoft SQL Server / T-SQL)
> **Project:** Telecom Customer Analytics & Churn Analysis System
> **Output:** `02_insert_data.sql` — T-SQL INSERT statements for all tables
>
> **Execution order:**
> 1. Run `01_create_database.sql` in SSMS or sqlcmd (creates schema + tables, no FK constraints)
> 2. Run this notebook to generate `02_insert_data.sql`
> 3. Run `02_insert_data.sql` to load all data
> 4. Uncomment and run the FK `ALTER TABLE` block at the bottom of `01_create_database.sql`

## 0. Setup & Imports

````python
import pandas as pd
import numpy as np
from pathlib import Path

# ── Configuration ─────────────────────────────────────────────────────────────
CSV_DIR    = Path('.')                # folder containing the 4 CSV files
OUTPUT_SQL = Path('02_insert_data.sql')
SCHEMA     = 'telecom'               # must match 01_create_database.sql
DB_NAME    = 'telecom_churn_db'      # must match the USE statement in 01_
BATCH_SIZE = 500                     # rows per INSERT VALUES block

print('Imports OK')
````

## 1. Load CSVs

````python
df_main = pd.read_csv(CSV_DIR / 'Telco_customer_churn.csv')
df_demo = pd.read_csv(CSV_DIR / 'Telco_customer_churn_demographics.csv')
df_loc  = pd.read_csv(CSV_DIR / 'Telco_customer_churn_location.csv')
df_svc  = pd.read_csv(CSV_DIR / 'Telco_customer_churn_services.csv')

# Normalise column names: strip leading/trailing whitespace
for df in [df_main, df_demo, df_loc, df_svc]:
    df.columns = df.columns.str.strip()

print(f'main : {df_main.shape}')
print(f'demo : {df_demo.shape}')
print(f'loc  : {df_loc.shape}')
print(f'svc  : {df_svc.shape}')
````

## 2. Utility Functions (T-SQL compatible)

````python
def sql_val(v):
    """
    Convert a Python value to a safe T-SQL literal.
    MSSQL differences vs PostgreSQL:
      - BOOLEAN does not exist in MSSQL; BIT columns use 1 / 0
      - No TRUE / FALSE keywords; use 1 / 0
      - NULL stays NULL
    """
    if v is None or (isinstance(v, float) and np.isnan(v)):
        return 'NULL'
    # BIT: Python bool maps to 1 / 0
    if isinstance(v, bool):
        return '1' if v else '0'
    if isinstance(v, (int, np.integer)):
        return str(int(v))
    if isinstance(v, (float, np.floating)):
        return str(float(v))
    # NVARCHAR string — escape single quotes by doubling them
    return "N'" + str(v).replace("'", "''") + "'"


def to_bool(val, yes_values=('Yes', 'yes', '1', 1, True, 'TRUE', 'true')):
    """Map Yes/No or 1/0 to Python bool (rendered as 1/0 by sql_val)."""
    if pd.isna(val):
        return False
    return val in yes_values


def safe_float(val, default=0.0):
    try:
        return float(val)
    except (ValueError, TypeError):
        return default


def batch_insert(table, columns, rows, batch_size=BATCH_SIZE):
    """
    Generate batched T-SQL INSERT INTO ... VALUES (...), (...) statements.
    MSSQL supports multi-row VALUES since SQL Server 2008.
    Each batch is followed by GO to flush the transaction log.
    """
    lines = []
    col_str = ', '.join(columns)
    for i in range(0, len(rows), batch_size):
        chunk = rows[i : i + batch_size]
        vals  = ',\n    '.join(
            '(' + ', '.join(sql_val(c) for c in row) + ')'
            for row in chunk
        )
        lines.append(
            f'INSERT INTO {SCHEMA}.{table} ({col_str}) VALUES\n    {vals};'
        )
    return '\nGO\n'.join(lines) + '\nGO\n'


print('Utility functions ready')
````

## 3. Build SQL Blocks

### 3.1 — `customer` (from demographics CSV)

````python
customer_rows = []
for _, r in df_demo.iterrows():
    cid = str(r['Customer ID']).strip()
    customer_rows.append((
        cid,                                     # customer_id      NVARCHAR
        str(r['Gender']).strip(),                # gender           NVARCHAR
        int(r['Age']),                           # age              INT
        to_bool(r.get('Under 30', 'No')),        # under_30         BIT -> 1/0
        to_bool(r['Senior Citizen']),            # senior_citizen   BIT -> 1/0
        to_bool(r['Married']),                   # married          BIT -> 1/0
        to_bool(r['Dependents']),                # dependents       BIT -> 1/0
        int(r['Number of Dependents']),          # no_dependents    INT
        None,                                    # referred_by      NULL (self-ref FK applied post-load)
    ))

sql_customer = batch_insert(
    'customer',
    ['customer_id', 'gender', 'age', 'under_30', 'senior_citizen',
     'married', 'dependents', 'no_dependents', 'referred_by'],
    customer_rows
)
print(f'customer rows: {len(customer_rows)}')
````

### 3.2 — `location` (from location CSV)

````python
loc_rows = []
for _, r in df_loc.iterrows():
    loc_rows.append((
        str(r['Location ID']).strip(),   # location_id
        str(r['Customer ID']).strip(),   # customer_id
        str(r['Country']).strip(),       # country
        str(r['State']).strip(),         # state
        str(r['City']).strip(),          # city
        str(int(r['Zip Code'])),         # zip_code (stored as NVARCHAR)
        str(r['Lat Long']).strip(),      # lat_long
        float(r['Latitude']),            # latitude
        float(r['Longitude']),           # longitude
    ))

sql_location = batch_insert(
    'location',
    ['location_id', 'customer_id', 'country', 'state', 'city',
     'zip_code', 'lat_long', 'latitude', 'longitude'],
    loc_rows
)
print(f'location rows: {len(loc_rows)}')
````

### 3.3 — `quarter` (distinct values from services CSV)

````python
distinct_quarters = sorted(df_svc['Quarter'].dropna().unique())
quarter_rows = [(q, q) for q in distinct_quarters]  # (quarter_id, quarter_name)

sql_quarter = batch_insert(
    'quarter',
    ['quarter_id', 'quarter_name'],
    quarter_rows
)
print(f'quarter rows: {len(quarter_rows)} -> {distinct_quarters}')
````

### 3.4 — `subscription` (from services CSV)

````python
sub_rows = []
for _, r in df_svc.iterrows():
    cid = str(r['Customer ID']).strip()
    offer = None if pd.isna(r.get('Offer')) else str(r['Offer']).strip()
    sub_rows.append((
        'SUB-' + cid,                        # sub_id
        cid,                                 # customer_id
        str(r['Contract']).strip(),          # contract_type
        offer,                               # offer_details  (nullable)
        str(r['Payment Method']).strip(),    # payment_method
        to_bool(r['Paperless Billing']),     # paperless_billing_option  BIT
    ))

sql_subscription = batch_insert(
    'subscription',
    ['sub_id', 'customer_id', 'contract_type', 'offer_details',
     'payment_method', 'paperless_billing_option'],
    sub_rows
)
print(f'subscription rows: {len(sub_rows)}')
````

### 3.5 — `service` (from services CSV)

````python
svc_rows = []
for _, r in df_svc.iterrows():
    cid = str(r['Customer ID']).strip()
    internet_type = None if pd.isna(r.get('Internet Type')) else str(r['Internet Type']).strip()
    svc_rows.append((
        str(r['Service ID']).strip(),                           # service_id
        cid,                                                    # customer_id
        to_bool(r['Phone Service']),                            # phone_service          BIT
        safe_float(r['Avg Monthly Long Distance Charges']),     # avg_monthly_ld_charges DECIMAL
        to_bool(r['Multiple Lines']),                           # multiple_lines         BIT
        to_bool(r['Internet Service']),                         # internet_service       BIT
        internet_type,                                          # internet_type          NVARCHAR NULL
        int(r['Avg Monthly GB Download']),                      # avg_monthly_gb_download INT
        to_bool(r['Online Security']),                          # online_security        BIT
        to_bool(r['Online Backup']),                            # online_backup          BIT
        str(r['Device Protection Plan']).strip(),               # device_protection_plan NVARCHAR
        to_bool(r['Premium Tech Support']),                     # premium_tech_support   BIT
        to_bool(r['Streaming TV']),                             # streaming_tv           BIT
        to_bool(r['Streaming Movies']),                         # streaming_movies       BIT
        to_bool(r['Streaming Music']),                          # streaming_music        BIT
        to_bool(r['Unlimited Data']),                           # unlimited_data         BIT
    ))

sql_service = batch_insert(
    'service',
    ['service_id', 'customer_id', 'phone_service', 'avg_monthly_ld_charges',
     'multiple_lines', 'internet_service', 'internet_type', 'avg_monthly_gb_download',
     'online_security', 'online_backup', 'device_protection_plan', 'premium_tech_support',
     'streaming_tv', 'streaming_movies', 'streaming_music', 'unlimited_data'],
    svc_rows
)
print(f'service rows: {len(svc_rows)}')
````

### 3.6 — `behavior` (from services CSV)

````python
beh_rows = []
for _, r in df_svc.iterrows():
    cid = str(r['Customer ID']).strip()
    beh_rows.append((
        'BEH-' + cid,
        cid,
        int(r['Avg Monthly GB Download']),
        safe_float(r['Avg Monthly Long Distance Charges']),
    ))

sql_behavior = batch_insert(
    'behavior',
    ['behavior_id', 'customer_id', 'avg_monthly_gb_download', 'long_distance_usage_metrics'],
    beh_rows
)
print(f'behavior rows: {len(beh_rows)}')
````

### 3.7 — `billing` (from services CSV)

````python
bil_rows = []
for _, r in df_svc.iterrows():
    cid = str(r['Customer ID']).strip()
    bil_rows.append((
        'BIL-' + cid,
        cid,
        safe_float(r['Monthly Charge']),
        safe_float(r['Total Charges']),
        safe_float(r['Total Revenue']),
        safe_float(r['Total Refunds']),
        safe_float(r['Total Extra Data Charges']),
        safe_float(r['Total Long Distance Charges']),
        safe_float(r['Avg Monthly Long Distance Charges']),
    ))

sql_billing = batch_insert(
    'billing',
    ['billing_id', 'customer_id', 'monthly_charges', 'total_charges', 'total_revenue',
     'total_refunds', 'total_extra_data_charges', 'total_long_distance_charges',
     'avg_monthly_long_dist_charges'],
    bil_rows
)
print(f'billing rows: {len(bil_rows)}')
````

### 3.8 — `churn_report` (from main CSV)

````python
churn_rows = []
for _, r in df_main.iterrows():
    cid    = str(r['CustomerID']).strip()
    reason = None if pd.isna(r.get('Churn Reason')) else str(r['Churn Reason']).strip()
    churn_rows.append((
        'CHU-' + cid,
        str(r['Churn Label']).strip(),
        int(r['Churn Value']),
        int(r['Churn Score']),
        reason,
    ))

sql_churn_report = batch_insert(
    'churn_report',
    ['churn_id', 'churn_label', 'churn_value', 'churn_score', 'churn_reason'],
    churn_rows
)
print(f'churn_report rows: {len(churn_rows)}')
````

### 3.9 — `customer_loyalty` (services + main CSV for CLTV)

````python
cltv_map = df_main.set_index('CustomerID')['CLTV'].to_dict()

loy_rows = []
for _, r in df_svc.iterrows():
    cid     = str(r['Customer ID']).strip()
    quarter = str(r['Quarter']).strip()
    loy_rows.append((
        'LOY-' + cid,
        cid,
        'BIL-' + cid,
        quarter,
        int(cltv_map.get(cid, 0)),
        int(r['Tenure in Months']),
        int(r['Number of Referrals']),
        to_bool(r['Referred a Friend']),   # BIT: 1 / 0
    ))

sql_loyalty = batch_insert(
    'customer_loyalty',
    ['loyalty_id', 'customer_id', 'billing_id', 'quarter_id', 'cltv',
     'tenure_months', 'number_of_referrals', 'referral_status'],
    loy_rows
)
print(f'customer_loyalty rows: {len(loy_rows)}')
````

### 3.10 — Junction: `customer_churn_report`

````python
ccr_rows = []
for _, r in df_main.iterrows():
    cid = str(r['CustomerID']).strip()
    ccr_rows.append((cid, 'CHU-' + cid))

sql_ccr = batch_insert(
    'customer_churn_report',
    ['customer_id', 'churn_id'],
    ccr_rows
)
print(f'customer_churn_report rows: {len(ccr_rows)}')
````

### 3.11 — Junction: `customer_service`

````python
cs_rows = []
for _, r in df_svc.iterrows():
    cid = str(r['Customer ID']).strip()
    sid = str(r['Service ID']).strip()
    cs_rows.append((cid, sid))

sql_cs = batch_insert(
    'customer_service',
    ['customer_id', 'service_id'],
    cs_rows
)
print(f'customer_service rows: {len(cs_rows)}')
````

## 4. Assemble & Write `02_insert_data.sql`

````python
HEADER = f"""\
-- =============================================================================
-- FILE: 02_insert_data.sql
-- PROJECT: Telecom Customer Analytics & Churn Analysis System
-- DBMS: Microsoft SQL Server (T-SQL)
-- DESCRIPTION: Data injection -- ordered to satisfy implicit FK relationships.
--              FK constraints are NOT active during this load; they are defined
--              in the commented block at the bottom of 01_create_database.sql.
--
-- INJECTION ORDER (safe dependency chain):
--   1. customer              (no deps)
--   2. location              (needs customer)
--   3. quarter               (no deps)
--   4. subscription          (needs customer)
--   5. service               (needs customer)
--   6. behavior              (needs customer)
--   7. billing               (needs customer)
--   8. churn_report          (no deps)
--   9. customer_loyalty      (needs customer, billing, quarter)
--  10. customer_churn_report (needs customer, churn_report)
--  11. customer_service      (needs customer, service)
-- =============================================================================

USE {DB_NAME};
GO

BEGIN TRANSACTION;
GO
"""

FOOTER = """\
COMMIT TRANSACTION;
GO
-- =============================================================================
-- END OF FILE: 02_insert_data.sql
-- After this file completes successfully, run the FK ALTER TABLE block
-- at the bottom of 01_create_database.sql to enforce referential integrity.
-- =============================================================================
"""

def section(title):
    bar = '-' * 77
    return f'\n-- {bar}\n-- {title}\n-- {bar}\n'

sql_file_content = (
    HEADER
    + section('1. customer')
    + sql_customer
    + section('2. location')
    + sql_location
    + section('3. quarter')
    + sql_quarter
    + section('4. subscription')
    + sql_subscription
    + section('5. service')
    + sql_service
    + section('6. behavior')
    + sql_behavior
    + section('7. billing')
    + sql_billing
    + section('8. churn_report')
    + sql_churn_report
    + section('9. customer_loyalty')
    + sql_loyalty
    + section('10. customer_churn_report (junction)')
    + sql_ccr
    + section('11. customer_service (junction)')
    + sql_cs
    + FOOTER
)

with open(OUTPUT_SQL, 'w', encoding='utf-8') as f:
    f.write(sql_file_content)

size_kb = OUTPUT_SQL.stat().st_size / 1024
print(f'Written: {OUTPUT_SQL}  ({size_kb:.1f} KB)')
````

## 5. Validation Summary

````python
counts = {
    'customer'              : len(customer_rows),
    'location'              : len(loc_rows),
    'quarter'               : len(quarter_rows),
    'subscription'          : len(sub_rows),
    'service'               : len(svc_rows),
    'behavior'              : len(beh_rows),
    'billing'               : len(bil_rows),
    'churn_report'          : len(churn_rows),
    'customer_loyalty'      : len(loy_rows),
    'customer_churn_report' : len(ccr_rows),
    'customer_service'      : len(cs_rows),
}

print(f'{"Table":<30} {"Rows":>8}')
print('-' * 40)
total = 0
for t, n in counts.items():
    print(f'{t:<30} {n:>8,}')
    total += n
print('-' * 40)
print(f'{"TOTAL":<30} {total:>8,}')
````

## 6. Quick Data Quality Checks

````python
# Check: every Customer ID in location exists in customer
loc_cids  = set(df_loc['Customer ID'].str.strip())
demo_cids = set(df_demo['Customer ID'].str.strip())
print(f'Location rows with unknown CustomerID : {len(loc_cids - demo_cids)}')

# Check: every Customer ID in services exists in customer
svc_cids = set(df_svc['Customer ID'].str.strip())
print(f'Service rows with unknown CustomerID  : {len(svc_cids - demo_cids)}')

# Check: every CustomerID in main exists in customer
main_cids = set(df_main['CustomerID'].str.strip())
print(f'Main rows with unknown CustomerID     : {len(main_cids - demo_cids)}')

# Check: null churn reasons (expected ~5174 = non-churners)
null_reasons = df_main['Churn Reason'].isna().sum()
print(f'Null Churn Reason (non-churners)      : {null_reasons}')

# Check: BIT column sanity (no unexpected values)
bit_cols = ['Phone Service', 'Multiple Lines', 'Internet Service',
            'Online Security', 'Online Backup', 'Premium Tech Support',
            'Streaming TV', 'Streaming Movies', 'Streaming Music',
            'Unlimited Data', 'Paperless Billing', 'Referred a Friend']
print('\nBIT column unique value check (services CSV):')
for col in bit_cols:
    if col in df_svc.columns:
        print(f'  {col}: {sorted(df_svc[col].dropna().unique().tolist())}')

print('\nQuality check complete -- 0 orphans expected above')
````


**Key MSSQL differences applied in this notebook vs PostgreSQL version:**

| Concern | PostgreSQL | This notebook (MSSQL) |
|---------|-----------|----------------------|
| Boolean literals | `TRUE` / `FALSE` | `1` / `0` (BIT column) |
| String prefix | `'...'` | `N'...'` (Unicode NVARCHAR) |
| Transaction | `BEGIN;` / `COMMIT;` | `BEGIN TRANSACTION;` / `COMMIT TRANSACTION;` |
| Batch separator | not needed | `GO` after each INSERT batch |
| Schema switch | `SET search_path` | `USE database; GO` |
| Restrict FK action | `ON DELETE RESTRICT` | `ON DELETE NO ACTION` |
