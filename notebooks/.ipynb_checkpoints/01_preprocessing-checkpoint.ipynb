# Telco Customer Segmentation — 01. Preprocessing Pipeline

**Goal:** Prepare the `Telco-Customer-Churn` dataset for unsupervised clustering.

This notebook is preprocessing-only (no EDA, no modeling). It builds a reusable
`scikit-learn` `ColumnTransformer` pipeline that:

1. Loads the raw CSV
2. Fixes data type issues (`TotalCharges`)
3. Drops non-informative / leakage-prone columns (`customerID`, `Churn`)
4. Encodes categorical features (One-Hot)
5. Scales numeric features (StandardScaler)
6. Fits the pipeline and exports:
   - `preprocessing_pipeline.joblib` — the fitted sklearn pipeline (reusable on new data)
   - `X_preprocessed.npy` — the transformed feature matrix ready for clustering
   - `feature_names.joblib` — the output feature names (for interpreting clusters later)
   - `customer_ids.csv` — customerID + raw Churn label kept aside (not used in clustering,
     only for later profiling/validation of clusters)

The **modeling notebook** (`02_modeling_clustering.ipynb`) picks up exactly where this
notebook leaves off, loading these artifacts instead of re-doing preprocessing.


## 1. Imports & config

```python
import pandas as pd
import numpy as np
import joblib
from pathlib import Path

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

pd.set_option("display.max_columns", 50)

DATA_PATH = Path("/mnt/user-data/uploads/WA_Fn-UseC_-Telco-Customer-Churn.csv")
OUTPUT_DIR = Path("/mnt/user-data/outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
```

## 2. Load raw data

```python
df_raw = pd.read_csv(DATA_PATH)
print(df_raw.shape)
df_raw.head()
```

## 3. Fix data types

`TotalCharges` is loaded as a string because 11 rows contain a blank string
(`" "`) instead of a number — these correspond to customers with `tenure == 0`
(brand-new customers who have not been billed yet). We coerce to numeric and
impute those rows with `0`, which is the correct value given zero tenure.


```python
df = df_raw.copy()

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

n_missing = df["TotalCharges"].isna().sum()
print(f"Rows with non-numeric TotalCharges (coerced to NaN): {n_missing}")
assert (df.loc[df["TotalCharges"].isna(), "tenure"] == 0).all(), \
    "Unexpected: missing TotalCharges rows should all have tenure == 0"

df["TotalCharges"] = df["TotalCharges"].fillna(0.0)

# SeniorCitizen is 0/1 but semantically categorical -> make it explicit Yes/No
# so it is treated consistently with the other binary categorical columns.
df["SeniorCitizen"] = df["SeniorCitizen"].map({0: "No", 1: "Yes"})

df.dtypes
```

## 4. Separate identifiers / target from features

- `customerID` is a unique key, not a feature -> excluded from clustering.
- `Churn` is the historical outcome label. Since this is an **unsupervised**
  segmentation task, `Churn` is **not** used as a clustering input (that would
  leak the label into the feature space). We keep it aside so the modeling
  notebook can use it later purely to *profile/validate* the resulting
  clusters (e.g. "does Cluster 3 have a higher churn rate?").


```python
id_cols = ["customerID"]
target_col = "Churn"

customer_ref = df[id_cols + [target_col]].copy()
customer_ref.to_csv(OUTPUT_DIR / "customer_ids.csv", index=False)

X = df.drop(columns=id_cols + [target_col])
print("Feature matrix shape (pre-transform):", X.shape)
X.head()
```

## 5. Define column groups

- **Numeric:** `tenure`, `MonthlyCharges`, `TotalCharges`
- **Categorical:** everything else (all binary / low-cardinality string columns)


```python
numeric_features = ["tenure", "MonthlyCharges", "TotalCharges"]
categorical_features = [c for c in X.columns if c not in numeric_features]

print("Numeric features:", numeric_features)
print(f"Categorical features ({len(categorical_features)}):", categorical_features)
```

## 6. Build the preprocessing pipeline

- Numeric branch: median imputation (defensive, in case of future missing
  values) + `StandardScaler` (clustering algorithms like K-Means are
  distance-based and require scaled features).
- Categorical branch: most-frequent imputation + `OneHotEncoder`
  (`drop="if_binary"` keeps true binary columns like `gender`/`Partner` as a
  single 0/1 column instead of two redundant one-hot columns; multi-category
  columns like `InternetService`, `Contract`, `PaymentMethod` are fully
  one-hot encoded).


```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="if_binary", handle_unknown="ignore")),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features),
    ],
    remainder="drop",
)

preprocessing_pipeline = Pipeline(steps=[("preprocessor", preprocessor)])
preprocessing_pipeline
```

## 7. Fit & transform

```python
X_transformed = preprocessing_pipeline.fit_transform(X)

# OneHotEncoder can return a sparse matrix -> densify for clustering algorithms
if hasattr(X_transformed, "toarray"):
    X_transformed = X_transformed.toarray()

feature_names = preprocessing_pipeline.named_steps["preprocessor"].get_feature_names_out()

print("Transformed feature matrix shape:", X_transformed.shape)
print("Number of output features:", len(feature_names))
```

```python
X_df = pd.DataFrame(X_transformed, columns=feature_names)
X_df.head()
```

## 8. Sanity checks

Quick structural checks before exporting — not exploratory analysis, just
verifying the pipeline did what it should.


```python
assert X_transformed.shape[0] == df.shape[0], "Row count changed during preprocessing!"
assert not np.isnan(X_transformed).any(), "NaNs remain after preprocessing!"
assert np.isfinite(X_transformed).all(), "Non-finite values found after preprocessing!"

print("Row count matches original data:", X_transformed.shape[0] == df.shape[0])
print("No NaNs:", not np.isnan(X_transformed).any())
print("All finite:", np.isfinite(X_transformed).all())
```

## 9. Export artifacts

Everything the modeling notebook needs to run clustering, with no dependency
on re-running this notebook.


```python
joblib.dump(preprocessing_pipeline, OUTPUT_DIR / "preprocessing_pipeline.joblib")
joblib.dump(list(feature_names), OUTPUT_DIR / "feature_names.joblib")
np.save(OUTPUT_DIR / "X_preprocessed.npy", X_transformed)

print("Saved:")
for f in ["preprocessing_pipeline.joblib", "feature_names.joblib",
          "X_preprocessed.npy", "customer_ids.csv"]:
    p = OUTPUT_DIR / f
    print(f" - {p}  ({p.stat().st_size:,} bytes)")
```
